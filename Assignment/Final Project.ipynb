```python
from bs4 import BeautifulSoup
import pandas as pd
with open("C:/Users/chieloka.eze/Desktop/HTML/SomeData.html") as fd:
    soup = BeautifulSoup(fd, "html.parser")
    all_table, l_header = soup.find_all("table"), []
    header = all_table[1].find("tr")
    for item in header:
        item = item.getText().strip()
        if item != "":
            l_header.append(item)
    print(l_header)

    html_data, data = all_table[1].find_all("tr")[1:], []
    for item in html_data:
        sub_data = []
        for sub_item in item:
            sub_item = sub_item.getText().strip()
            if sub_item != "":
                sub_data.append(sub_item)
        data.append(sub_data)
    print(data)

df = pd.DataFrame(data = data, columns = l_header)
df
    
```
