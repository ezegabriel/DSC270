```python
cut = {}
for i in df["cut"]:
    if not i in cut:
        cut[i] = 1
    else:
        cut[i] += 1
cut
```

```python
cut["Ideal"], l_key = 2000, []
for j in cut.keys():
    if j == "Fair":
        l_key.append(j)
    elif j == "Ideal":
        l_key.append(j)
    elif j == "Premium":
        l_key.append(j)
l_key.sort()

for key in l_key:
    print(key, ": ", cut[key], sep = "")
```

```python
from bs4 import BeautifulSoup
import pandas as pd

with open("diamonds.html") as fd:
    soup = BeautifulSoup(fd, "html.parser")
    all_list, l_2d = soup.find_all("ol")[2], []
    
    for item in all_list:
        data, item = [], item.get_text().strip()
        data.append(item)
        if not "" in data:
            l_2d.append(item.split(","))
            
    print(l_2d)
    
    l_transpose = []
    for i_col in range(len(l_2d[0])):
        sub_data = []
        for i_row in range(len(l_2d)):
            sub_data.append(l_2d[i_row][i_col])   
        l_transpose.append(sub_data)
        
    print(l_transpose)
```
